## 1. Differential-rotation profiles

Presents the main latitude-dependent rotation profiles, fitted curves, and comparisons among the analyzed Ca II K data products and time intervals.

**Inputs:** `data/diff_rot.sav`, `data/asym_fit.sav`, `data/diff_rotall.sav`, `data/diff_allcak_new1.sav`, `data/diff_rot_1980_after.sav`, `data/diff_rot_1980_before.sav`  
**Output:** `figures/finalresult_new.pdf`

```python
# import libraries
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import scipy
from os.path import dirname, join as pjoin
import scipy.io as sio
from scipy.io import readsav
from math import sin
mpl.style.use('config/bkj_style.mplstyle')

# read 'sav' files
sav_data1 = readsav("data/diff_rot.sav")
sav_data4 = readsav("data/asym_fit.sav")
sav_data5 = readsav("data/diff_rotall.sav")
sav_data6 = readsav("data/diff_allcak_new1.sav")
sav_data7 = readsav("data/diff_rot_1980_after.sav")
sav_data8 = readsav("data/diff_rot_1980_before.sav")


# plot figure
fig=plt.figure(figsize = (6,6))
ax1 = fig.add_axes([0.1,0.1,0.35, 0.35])
ax2 = fig.add_axes([0.46,0.1,0.35, 0.35])
ax3 = fig.add_axes([0.82,0.1,0.35, 0.35])


# rotation profile comparision plot

ax1.plot(sav_data7['ll'],sav_data7['yy4'],color='purple',label = 'Ca-K (KoSO)${^1}$',linestyle='--')
ax1.plot(sav_data1['lat'],sav_data1['omega'],'o',color='red',markersize=1)
ax1.set_ylim(12,15.0)
ax1.set_xlim(-60,60)
ax1.set_xlabel("Latitude [$^{\circ}$]",fontsize=10)
ax1.set_ylabel("$\Omega$ [$^{\circ}$ /day]",fontsize=10)
ax1.tick_params(axis='both',labelsize=10)


ax1.fill_between(sav_data1['lat'],sav_data1['omega']-sav_data1['std'],sav_data1['omega']+sav_data1['std'],color='red', alpha=0.07)
ax1.plot(sav_data1['ll'],sav_data1['yy4'],color='red',label = 'Ca-K (KoSO)$^{2}$')
ax1.plot(sav_data8['ll'],sav_data8['yy4'],color='olive',label = 'Ca-K (KoSO)$^{3}$',linestyle='dashdot')
ax1.plot(sav_data1['ll'],sav_data1['yy1'],color='blue',label = 'WL',linestyle='--')
ax1.set_xticks([-50,-30,-10,10,30,50])
ax1.annotate('a)',(-55,14.6))
ax1.legend(fontsize=7.5,loc=(0.3,0.23),frameon= False,labelcolor='linecolor',handlelength=1.4)



# rotation profile other results plot
ax2.plot(sav_data5['ll'],sav_data5['yy2'],color='darkorange',label ='Wan+ 2022',linestyle=(0,(5,1)))
ax2.plot(sav_data5['ll'],sav_data5['yy6'],color='darkslategrey',label ='Antonucci+ 1979',linestyle='dashdot')
ax2.plot(sav_data5['lat'],sav_data5['omega'],'o',color='red',markersize=1)
ax2.set_ylim(12,15.0)
ax2.set_xlim(-60,60)
ax2.set_xlabel("Latitude [$^{\circ}$]",fontsize=10)
ax2.tick_params(axis='both',labelsize=10,labelleft=False)
ax2.fill_between(sav_data5['lat'],sav_data5['omega']-sav_data5['std'],sav_data5['omega']+sav_data5['std'],color='red', alpha=0.05)
ax2.plot(sav_data5['ll'],sav_data5['yy4'],color='red',label = 'Ca-K (KoSO)')

ax2.plot(sav_data5['ll'],sav_data5['yy1'],color='blue',label = 'WL',linestyle='--')
ax2.plot(sav_data1['ll'],sav_data1['yy2'],color='brown',label ='Bertello+ 2020',linestyle='dashdot')
ax2.plot(sav_data1['ll'],sav_data1['yy3'],color='mediumvioletred',label ='Howard+ 1983',linestyle =(0,(1,1)))
ax2.set_xticks([-50,-30,-10,10,30,50])
ax2.annotate('b)',(-55,14.6))
ax2.legend(fontsize=7.5,loc=(0.26,0.1),frameon= False,labelcolor='linecolor',handlelength=1.4)

# rotation profile for all other ca k data
ax3.plot(sav_data6['lat'],sav_data6['omega'],'o',color='red',markersize=1)
ax3.set_ylim(12,15.0)
ax3.set_xlim(-60,60)
ax3.set_xlabel("Latitude [$^{\circ}$]",fontsize=10)
ax3.tick_params(axis='both',labelsize=10,labelleft=False)

ax3.fill_between(sav_data6['lat'],sav_data6['omega']-sav_data6['std'],sav_data6['omega']+sav_data6['std'],color='red', alpha=0.05)
ax3.plot(sav_data6['ll'],sav_data6['yy4'],color='red',label = 'KoSO')
ax3.plot(sav_data6['_lat'],sav_data6['_omega'],'o',color='teal',markersize=1)
ax3.fill_between(sav_data6['_lat'],sav_data6['_omega']-sav_data6['_std'],sav_data6['_omega']+sav_data6['_std'],color='teal', alpha=0.05)
ax3.plot(sav_data6['ll'],sav_data6['yy8'],color='teal',label = 'Rome/PSPT')
ax3.plot(sav_data6['lat_'],sav_data6['omega_'],'o',color='purple',markersize=1)
ax3.fill_between(sav_data6['lat_'],sav_data6['omega_']-sav_data6['std_'],sav_data6['omega_']+sav_data6['std_'],color='purple', alpha=0.05)
ax3.plot(sav_data6['ll'],sav_data6['yy7'],color='purple',label = 'Meudon')
ax3.plot(sav_data6['ll'],sav_data6['yy1'],color='blue',label ='WL',linestyle='--')
ax3.set_xticks([-50,-30,-10,10,30,50])
ax3.annotate('c)',(-55,14.6))
ax3.legend(fontsize=7.5,loc=(0.34,0.22),frameon= False,labelcolor='linecolor',handlelength=1.4)

# save figures in "pdf" format
plt.savefig("figures/finalresult_new.pdf",dpi=300,bbox_inches='tight')
plt.show()
```

## 2. Symmetric and hemispheric rotation profiles

Compares the symmetric full-disk fit with an asymmetric fit and with rotation profiles derived independently for the northern and southern hemispheres.

**Inputs:** `data/diff_rot.sav`, `data/north_hem.sav`, `data/south_hem.sav`, `data/asym_fit.sav`  
**Output:** `figures/result_ns.pdf`

```python
# import libraries
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import scipy
from os.path import dirname, join as pjoin
import scipy.io as sio
from scipy.io import readsav
from math import sin
mpl.style.use('config/bkj_style.mplstyle')

# read 'sav' files
sav_data1 = readsav("data/diff_rot.sav")
sav_data2 = readsav("data/north_hem.sav")
sav_data3 = readsav("data/south_hem.sav")
sav_data4 = readsav("data/asym_fit.sav")

# plot figure
fig=plt.figure(figsize = (6,6))
ax2 = fig.add_axes([0.1,0.1,0.45, 0.4])

# rotation profile for symmetric and asymmetric fitting
ax2.plot(sav_data1['lat'],sav_data1['omega'],'o',color='yellow',markersize=1)
ax2.errorbar(sav_data1['lat'],sav_data1['omega'],yerr=sav_data1['std'],fmt='.k',capsize=1,capthick=0.5,color='orange',markersize=1,elinewidth=0.5)
ax2.plot(sav_data1['ll'],sav_data1['yy4'],color='orange',label = 'Symmetric Profile',linestyle='--')
ax2.set_xticks([-50,-30,-10,10,30,50])


ax2.plot(sav_data4['lat'],sav_data4['omega'],'o',color='hotpink',markersize=1)
ax2.set_ylim(12,15.0)
ax2.set_xlabel("Latitude [deg]",fontsize=10)
ax2.tick_params(axis='both',labelsize=10)
ax2.errorbar(sav_data4['lat'],sav_data4['omega'],yerr=sav_data4['std'],fmt='.k',capsize=1,capthick=0.5,color='hotpink',markersize=1,elinewidth=0.5)
ax2.plot(sav_data4['ll'],sav_data4['yy4'],color='hotpink',label = 'Asymmetric Profile',linestyle='dashdot')


# rotation profile for north and south hemisphere
lat2 = sav_data2['lat2']
lat2_ = lat2[::-1]
omega_s = sav_data3['omega_s']
omega_s_ = omega_s[::-1]
std2 = sav_data3['std2']
std2_ = std2[::-1]
yy4 = sav_data3['yy4']
yy4_ = yy4[::-1]
ll = sav_data3['ll']
ll_ = ll[::-1]

ax2.plot(sav_data2['lat2'],sav_data2['omega_n'],'o',color='teal',markersize=1)
ax2.set_ylim(12,15.0)
ax2.set_xlabel("Latitude [$^{\circ}$]",fontsize=10)
ax2.set_ylabel("$\Omega$ [$^{\circ}$ /day]",fontsize=10)
ax2.tick_params(axis='both',labelsize=10)
ax2.errorbar(sav_data2['lat2'],sav_data2['omega_n'],yerr=sav_data2['std2'],fmt='.k',capsize=1,capthick=0.5,color='teal',markersize=1,elinewidth=0.5)
ax2.plot(sav_data2['ll'],sav_data2['yy4'],color='teal',label = 'Northern Hemisphere')

ax2.plot(np.negative(lat2_),omega_s_,'o',color='dodgerblue',markersize=1)
ax2.errorbar(np.negative(lat2_),omega_s_,yerr=std2_,fmt='.k',capsize=1,capthick=0.5,color='dodgerblue',markersize=1,elinewidth=0.5)
ax2.plot(np.negative(ll_),yy4_,color='dodgerblue',label = 'Southern Hemisphere')
ax2.legend(fontsize=9,loc=(0.25,0.25),frameon= False,labelcolor='linecolor')



# save figures in "pdf" format
plt.savefig("figures/result_ns.pdf",dpi=300,bbox_inches='tight')
plt.show()
```

## 3. Mount Wilson calibration tests

Compares Mount Wilson Observatory rotation profiles produced with alternative calibration and correction choices, including distortion and limb-darkening treatments.

**Inputs:** `data/diff_wilson_new.sav`, `data/diff_wilson_new1.sav`, `data/diff_wilson_elip_corrected.sav`, `data/diff_wilson_elip_notcorrected.sav`, `data/diff_wilson_wo_ldc.sav`  
**Output:** `figures/result_mwo_all.pdf`

```python
# import libraries
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import scipy
from os.path import dirname, join as pjoin
import scipy.io as sio
from scipy.io import readsav
from math import sin
mpl.style.use('config/bkj_style.mplstyle')

# read 'sav' files
sav_data1 = readsav("data/diff_wilson_new.sav")
sav_data2 = readsav("data/diff_wilson_new1.sav")
sav_data3 = readsav("data/diff_wilson_elip_corrected.sav")
sav_data4 = readsav("data/diff_wilson_elip_notcorrected.sav")
sav_data5 = readsav("data/diff_wilson_wo_ldc.sav")



# plot figure
fig=plt.figure(figsize = (6,6))
ax1 = fig.add_axes([0.1,0.1,0.45, 0.4])

ax1.plot(sav_data1['lat_'],sav_data1['omega_'],'o',color='darkgreen',markersize=1)
ax1.set_ylim(12.0,15.0)
ax1.set_xlabel("Latitude [$^{\circ}$]",fontsize=10)
ax1.set_ylabel("$\Omega$ [$^{\circ}$ / day]",fontsize=10)
ax1.tick_params(axis='both',labelsize=10)
ax1.fill_between(sav_data1['lat_'],sav_data1['omega_']-sav_data1['std_'],sav_data1['omega_']+sav_data1['std_'],color='darkgreen', alpha=0.05)
ax1.plot(sav_data1['ll'],sav_data1['yy7'],color='darkgreen',label = 'Calibrated')

ax1.plot(sav_data4['lat'],sav_data4['omega'],'o',color='orchid',markersize=1)
ax1.set_ylim(12.0,15.0)
ax1.set_xlabel("Latitude [$^{\circ}$]",fontsize=10)
ax1.set_ylabel("$\Omega$ [$^{\circ}$ / day]",fontsize=10)
ax1.tick_params(axis='both',labelsize=10)
ax1.fill_between(sav_data4['lat'],sav_data4['omega']-sav_data4['std'],sav_data4['omega']+sav_data4['std'],color='orchid', alpha=0.05)
ax1.plot(sav_data4['ll'],sav_data4['yy4'],color='orchid',label = 'Raw without DC')


ax1.plot(sav_data3['lat'],sav_data3['omega'],'o',color='orange',markersize=1)
ax1.set_ylim(12.0,15.0)
ax1.set_xlabel("Latitude [$^{\circ}$]",fontsize=10)
ax1.set_ylabel("$\Omega$ [$^{\circ}$ / day]",fontsize=10)
ax1.tick_params(axis='both',labelsize=10)
ax1.fill_between(sav_data3['lat'],sav_data3['omega']-sav_data3['std'],sav_data3['omega']+sav_data3['std'],color='orange', alpha=0.05)
ax1.plot(sav_data3['ll'],sav_data3['yy4'],color='orange',label = 'Raw with DC')

ax1.plot(sav_data2['lat_'],sav_data2['omega_'],'o',color='darkmagenta',markersize=1)
ax1.set_ylim(12.0,15.0)
ax1.set_xlabel("Latitude [$^{\circ}$]",fontsize=10)
ax1.set_ylabel("$\Omega$ [$^{\circ}$ / day]",fontsize=10)
ax1.tick_params(axis='both',labelsize=10)
ax1.fill_between(sav_data2['lat_'],sav_data2['omega_']-sav_data2['std_'],sav_data2['omega_']+sav_data2['std_'],color='darkmagenta', alpha=0.05)
ax1.plot(sav_data2['ll'],sav_data2['yy7'],color='darkmagenta',label = 'Calibrated')


ax1.plot(sav_data5['lat'],sav_data5['omega'],'o',color='indianred',markersize=1)
ax1.set_ylim(12.0,15.0)
ax1.set_xlabel("Latitude [$^{\circ}$]",fontsize=10)
ax1.set_ylabel("$\Omega$ [$^{\circ}$ / day]",fontsize=10)
ax1.tick_params(axis='both',labelsize=10)
ax1.fill_between(sav_data5['lat'],sav_data5['omega']-sav_data5['std'],sav_data5['omega']+sav_data5['std'],color='indianred', alpha=0.05)
ax1.plot(sav_data5['ll'],sav_data5['yy4'],color='indianred',label = 'Raw')


ax1.set_xticks([-60,-40,-20,0,20,40,60])
ax1.legend(fontsize=7,bbox_to_anchor=(0.5, 0.49),frameon= False,labelcolor='linecolor',handlelength=1.0,labelspacing=0.9)
ax1.annotate('Chatzistergos+ 2020',(-12,13.4),fontsize=6)
ax1.annotate('Bertello+ 2010',(-12,12.72),fontsize=6)

# save figures in "pdf" format
plt.savefig("figures/result_mwo_all.pdf", format="pdf",dpi=300,bbox_inches='tight')
plt.show()
```

## 4. KoSO–MWO comparison

Compares the KoSO and Mount Wilson differential-rotation profiles. The scatter panel reports linear fits and Pearson correlation coefficients for the paired measurements.

**Inputs:** `data/diff_wilson_new.sav`, `data/wilson_scatter_new.sav`, `data/diff_wilson_new1.sav`, `data/wilson_scatter_new1.sav`  
**Output:** `figures/result2_new.pdf`

```python
# import libraries
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import scipy
from os.path import dirname, join as pjoin
import scipy.io as sio
from scipy.io import readsav
from math import sin
from scipy.stats import pearsonr
mpl.style.use('config/bkj_style.mplstyle')

# read 'sav' files
sav_data1 = readsav("data/diff_wilson_new.sav")
sav_data2 = readsav("data/wilson_scatter_new.sav")
sav_data3 = readsav("data/diff_wilson_new1.sav")
sav_data4 = readsav("data/wilson_scatter_new1.sav")

# plot figure
fig=plt.figure(figsize = (6,6))
ax1 = fig.add_axes([0.1,0.1,0.45, 0.4])
ax2 = fig.add_axes([0.65,0.1,0.42, 0.4])

# rotation profile comparision b/w koso and mwo data
ax1.plot(sav_data1['lat'],sav_data1['omega'],'o',color='red',markersize=1)
ax1.set_ylim(12.0,15.0)
ax1.set_xlabel("Latitude [$^{\circ}$]",fontsize=10)
ax1.set_ylabel("$\Omega$ [$^{\circ}$ / day]",fontsize=10)
ax1.tick_params(axis='both',labelsize=10)
ax1.fill_between(sav_data1['lat'],sav_data1['omega']-sav_data1['std'],sav_data1['omega']+sav_data1['std'],color='red', alpha=0.05)
ax1.plot(sav_data1['ll'],sav_data1['yy4'],color='red',label = 'This work (KoSO)')

ax1.plot(sav_data1['lat_'],sav_data1['omega_'],'o',color='darkgreen',markersize=1)
ax1.set_ylim(12.0,15.0)
ax1.set_xlabel("Latitude [$^{\circ}$]",fontsize=10)
ax1.set_ylabel("$\Omega$ [$^{\circ}$ / day]",fontsize=10)
ax1.tick_params(axis='both',labelsize=10)
ax1.fill_between(sav_data1['lat_'],sav_data1['omega_']-sav_data1['std_'],sav_data1['omega_']+sav_data1['std_'],color='darkgreen', alpha=0.05)
ax1.plot(sav_data1['ll'],sav_data1['yy7'],color='darkgreen',label = 'This work (MWO)$^{1}$')

ax1.plot(sav_data1['ll'],sav_data1['yy2'],color='brown',label ='Bertello+ 2020',linestyle='--')

ax1.plot(sav_data3['lat_'],sav_data3['omega_'],'o',color='darkmagenta',markersize=1)
ax1.set_ylim(12.0,15.0)
ax1.set_xlabel("Latitude [$^{\circ}$]",fontsize=10)
ax1.set_ylabel("$\Omega$ [$^{\circ}$ / day]",fontsize=10)
ax1.tick_params(axis='both',labelsize=10)
ax1.fill_between(sav_data3['lat_'],sav_data3['omega_']-sav_data3['std_'],sav_data3['omega_']+sav_data3['std_'],color='darkmagenta', alpha=0.05)
ax1.plot(sav_data3['ll'],sav_data3['yy7'],color='darkmagenta',label = 'This work (MWO)$^{2}$')

ax1.set_xticks([-60,-40,-20,0,20,40,60])
ax1.legend(fontsize=7,bbox_to_anchor=(0.5, 0.45),frameon= False,labelcolor='linecolor')
ax1.annotate('c)',(-57,14.75))

# scatter plot
ax2.scatter(sav_data2['omega'],sav_data2['omega_'],s=2,color='darkgreen')
ax2.scatter(sav_data4['omega'],sav_data4['omega_'],s=2,color='darkmagenta')
ax2.set_xlabel("$\Omega$$_{KoSO}$ [$^{\circ}$ / day]",fontsize=10)
ax2.set_ylabel("$\Omega$$_{MWO}$ [$^{\circ}$ / day]",fontsize=10)
ax2.tick_params(axis='both',labelsize=10)
el = ax2.errorbar(sav_data2['omega'],sav_data2['omega_'],xerr= sav_data2['std'],yerr=sav_data1['std_'],fmt='.k',capsize=1,capthick=0.5,ecolor='darkgreen',markersize=2,elinewidth=0.5)
el_ = ax2.errorbar(sav_data4['omega'],sav_data4['omega_'],xerr= sav_data4['std'],yerr=sav_data3['std_'],fmt='.k',capsize=1,capthick=0.5,ecolor='darkmagenta',markersize=2,elinewidth=0.5)

# compute slope m and intercept b
m, b = np.polyfit(sav_data2['omega'],sav_data2['omega_'], deg=1)
m_, b_ = np.polyfit(sav_data4['omega'],sav_data4['omega_'], deg=1)

# plot fitted y = m*x + b
ax2.plot(sav_data2['omega'], b + m * sav_data2['omega'], '-',label=f'$y = {m:.2f}x {b:+.2f}$',color='darkcyan')
ax2.plot(sav_data4['omega'], b_ + m_ * sav_data4['omega'], '-',label=f'$y = {m_:.2f}x {b_:+.2f}$',color='indigo')
ax2.legend(fontsize=8,bbox_to_anchor=(0.5, 0.25),frameon= False)

corr, _ = pearsonr(sav_data2['omega'],sav_data2['omega_'])
corr_, _ = pearsonr(sav_data4['omega'],sav_data4['omega_'])

ax2.annotate('CC = '+ "{:.2f}".format(corr),(13.6,14.0),fontsize=10)
ax2.annotate('CC = '+ "{:.2f}".format(corr_),(14.3,13.4),fontsize=10)
ax2.annotate('d)',(13.45,14.5))

# save figures in "pdf" format
plt.savefig("figures/result2_new.pdf", format="pdf",dpi=300,bbox_inches='tight')
plt.show()
```

## 5. MDI method validation

Validates the image-correlation method against feature tracking using MDI data. The second panel quantifies their agreement with a linear fit and Pearson correlation coefficient.

**Inputs:** `data/diff_mdi1.sav`, `data/mdi_scatter.sav`  
**Output:** `figures/result4.pdf`

```python
# import libraries
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import scipy
from os.path import dirname, join as pjoin
import scipy.io as sio
from scipy.io import readsav
from math import sin
from scipy.stats import pearsonr
mpl.style.use('config/bkj_style.mplstyle')

# read 'sav' files
sav_data1 = readsav("data/diff_mdi1.sav")
sav_data2 = readsav("data/mdi_scatter.sav")

# plot figure
fig=plt.figure(figsize = (6,6))
ax1 = fig.add_axes([0.1,0.1,0.42, 0.4])
ax2 = fig.add_axes([0.65,0.1,0.42, 0.4])

# rotation profile comparision of MDI data with different methods
ax1.plot(sav_data1['lat2'],sav_data1['omega_'],'s',color='navy',markersize=1.5,label = 'CI (MDI); Image Correlation')
ax1.set_ylim(13,14.8)
ax1.set_xlim(0,30)
ax1.set_xlabel("Latitude [$^{\circ}$]",fontsize=10)
ax1.set_ylabel("$\Omega$ [$^{\circ}$ / day]",fontsize=10)
ax1.tick_params(axis='both',labelsize=10)
ax1.errorbar(sav_data1['lat2'],sav_data1['omega_'],yerr=sav_data1['std2'],fmt='s',capsize=1,capthick=0.5,color='navy',markersize=1.5,elinewidth=0.5)

ax1.plot(sav_data1['_LLAT'],sav_data1['_OOMEGA'],'d',color='firebrick',markersize=1.5,label = 'CI (MDI); Tracking Method')
ax1.set_ylim(13,14.8)
ax1.set_xlabel("Latitude [$^{\circ}$]",fontsize=10)
ax1.set_ylabel("$\Omega$ [$^{\circ}$ / day]",fontsize=10)
ax1.tick_params(axis='both',labelsize=10)
ax1.errorbar(sav_data1['_LLAT'],sav_data1['_OOMEGA'],yerr=sav_data1['_ERR'],fmt='d',capsize=1,capthick=0.5,color='firebrick',markersize=1.5,elinewidth=0.5)

ax1.set_xticks([0,5,10,15,20,25,30])
ax1.legend(fontsize=8,bbox_to_anchor=(0.75, 0.25),frameon= False,labelcolor='linecolor')
ax1.annotate('a)',(1,14.62))

# scatter plot
ax2.scatter(sav_data2['omega'],sav_data2['_OOMEGA'],s=2)
ax2.set_xlim(13.8,14.5)
ax2.set_ylim(13.8,14.5)
ax2.set_xlabel("$\Omega$ $_{Image \ Correlation}$ [$^{\circ}$ / day]",fontsize=10)
ax2.set_ylabel("$\Omega$ $_{Tracking \ Method}$ [$^{\circ}$ / day]",fontsize=10)
ax2.tick_params(axis='both',labelsize=10)
ax2.set_yticks([13.8,14.0,14.2,14.4])
el = ax2.errorbar(sav_data2['omega'],sav_data2['_OOMEGA'],xerr= sav_data2['std'],yerr=sav_data2['_ERR'],fmt='.k',capsize=1,capthick=0.5,ecolor='darkgreen',markersize=2,elinewidth=0.5)


# compute slope m and intercept b
m, b = np.polyfit(sav_data2['omega'],sav_data2['_OOMEGA'], deg=1)

# plot fitted y = m*x + b
ax2.plot(sav_data2['omega'], b + m * sav_data2['omega'], '-',label=f'$y = {m:.2f}x {b:+.2f}$')
ax2.legend(fontsize=8,bbox_to_anchor=(0.5, 0.22),frameon= False)

corr, _ = pearsonr(sav_data2['omega'],sav_data2['_OOMEGA'])

ax2.annotate('CC = '+ "{:.2f}".format(corr),(13.9,14.2),fontsize=10)
ax2.annotate('b)',(13.82,14.43))

# save figures in "pdf" format
plt.savefig("figures/result4.pdf", format="pdf",dpi=300,bbox_inches='tight')
plt.show()
```

## 6. Long-term rotation-rate variation
